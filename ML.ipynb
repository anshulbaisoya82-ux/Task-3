```python
df['credit_risk_score'] = df['credit_risk_score'].replace(-1,np.nan)
df['credit_risk_score'] = df['credit_risk_score'].fillna(df['credit_risk_score'].mean())
print(f"{df['credit_risk_score'].isna().sum()} null values are in credit_risk_score")
df['current_address_months_count'] = df['current_address_months_count'].replace(-1,np.nan)
df['current_address_months_count'] = df['current_address_months_count'].fillna(df['current_address_months_count'].mean())
print(f"{df['current_address_months_count'].isna().sum()} null values are in current_address_months_count")
df['bank_months_count'] = df['bank_months_count'].replace(-1,np.nan)
df['bank_months_count'] = df['bank_months_count'].fillna(df['bank_months_count'].mean())
print(f"{df['bank_months_count'].isna().sum()} null values are in bank_months_count")
df['session_length_in_minutes'] = df['session_length_in_minutes'].replace(-1,np.nan)
df['session_length_in_minutes'] = df['session_length_in_minutes'].fillna(df['session_length_in_minutes'].mean())
print(f"{df['session_length_in_minutes'].isna().sum()} null values are in session_length_in_minutes")
df['device_distinct_emails_8w'] = df['device_distinct_emails_8w'].replace(-1,np.nan)
df['device_distinct_emails_8w'] = df['device_distinct_emails_8w'].fillna(df['device_distinct_emails_8w'].mean())
print(f"{df['device_distinct_emails_8w'].isna().sum()} null values are in device_distinct_emails_8w")
```
