# Part 2 (Dimensionality Reduction)

#### scaling data for pca

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
```

```python
encoder = OneHotEncoder(sparse_output=False)
```

```python
encoded = encoder.fit_transform(df[['payment_type','employment_status','housing_status','source','device_os']])
```

```python
encoded_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out(['payment_type','employment_status','housing_status','source','device_os']))
```

```python
encoded_df
```

```python
y = df.iloc[:,0]
```

```python
col = ['fraud_bool','payment_type','employment_status','housing_status','source','device_os']
```

```python
x = pd.concat([encoded_df,df.drop(columns=col)],axis=1)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2,random_state=69)
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
```

```python
x_test_scaled = scaler.transform(x_test)
```

```python
pca = PCA()
```

```python
pca.fit_transform(x_train_scaled)
pca.transform(x_test_scaled)
```

```python
explained_var = pca.explained_variance_ratio_
cumulative_var = np.cumsum(explained_var)
```

```python
plt.plot(range(1, len(explained_var) + 1), explained_var)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot')
```

```python
plt.plot(range(1, len(cumulative_var) + 1), cumulative_var)
plt.axhline(y=0.95, label='95% threshold')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Variance Plot')
plt.legend()
plt.show()
```

```python
n_components_95 = np.argmax(cumulative_var >= 0.95) + 1
n_components_90 = np.argmax(cumulative_var >= 0.90) + 1
print(f"Components needed for 95% variance: {n_components_95}")
print(f"Components needed for 90% variance: {n_components_90}")
```

- We have to select or extract features from our dataset due to curse of dimensity (higher the number of features, higher the computation and because of some no relational features in dataset, it decrease performance as well) because of high number of features present in a dataset. 
    So PCA is a method of feature extraction in which we create new features from old features to maximize the variance of that feature. Maximizing variance = more information. 

- In my dataset there are multiple features which are not providing much information about the fraud detection, so it's better to exclude them using feature extraction.

#### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score,accuracy_score,precision_score,f1_score,recall_score,confusion_matrix
```

```python
rf = RandomForestClassifier(n_estimators=200,class_weight='balanced',max_features=0.4,max_samples=0.7,random_state=69)
```

```python
print(f"roc_score:{roc_auc_score(y_test,y_pred)}")
print(f"accuracy_score {accuracy_score(y_test,y_pred)}")
print(f"precision_score:{precision_score(y_test,y_pred)}")
print(f"recall_score:{recall_score(y_test,y_pred)}")
print(f"f1_score:{f1_score(y_test,y_pred)}")
print(f"\n\nConfusion matrix is:\n {confusion_matrix(y_test,y_pred)}")
```
